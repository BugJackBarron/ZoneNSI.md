## Présentation

!!! info IMDB
Depuis 1990, le projet **Intenet Movie Database**, dont l'acronyme est **IMDb**, est une base de données en ligne sur le cinéma mondial, sur la télévision, et plus secondairement les jeux vidéo. IMDb restitue un grand nombre d’informations concernant les films, les acteurs, les réalisateurs, les scénaristes et toutes personnes et entreprises intervenant dans l’élaboration d’un film, d’un téléfilm, d’une série télévisée ou d’un jeu vidéo. 

IMDb est la propriété d’*Amazon*, mais son enrichissement et sa mise à jour sont ouverts, gratuitement, à toute personne disposant d’une adresse de courrier électronique et d’un accès à Internet, quelle que soit sa nationalité. **Les données sont exclusivement en anglais**. Le fondateur de la base de données et du site web éponyme, Col Needham a précisé que les informations ajoutées sont vérifiées par le site. 

Le fichier général d'IMDb est énorme (plusieurs Go), et est présenté sous la forme d'une base de données classique. Il est inutilisable en l'état au vu de vos connaissances actuelles.

J'en ai extrait deux fichiers au format `csv`, disponibles dans l'archive suivante : [TP_IMDB.zip](https://fvergniaud-drive.mytoutatice.cloud/public?sharecode=6pnjH9qFU9EB)

Vous devez télécharger le fichier, extraire les deux fichiers `csv`, puis les importer dans ce notebook (voir partie suivante).
